```python
import numpy as np
import pandas as pd
```

```python
data = {'Name': ['Asha','Ravi','Meera','Karan'],
        'Marks' : [88, 72, 91, 65],
        'Branch': ['CSE','ECE','CSE','ISE']}
df = pd.DataFrame(data)
print(df)
print("\nFirst 2 Rows :\n",df.iloc[0:2])
print("\nMarks column:\n",df['Marks'])
print("\nRows where marks >90:\n",df[df['Marks']>90])

```
